#Exercise 1: Prime Numbers

```python
num = int(input("Enter a number: ")) # Take input from the user

tag = False

if num == 0 or num == 1: # 0 and 1 is not prime
    print(num, "is not a prime number")
elif num > 1:
    # check for factors
    for i in range(2, num):
        if (num % i) == 0:
            # if factor is found, set tag to True
            tag = True
            # break out of loop
            break

    # check if flag is True
    if tag:
        print(num, "is not a prime number")
    else:
        print(num, "is a prime number")
```

#Exercise 3: Squares of Even/Odd Numbers


```python
# Ask the user for their choice: Even or Odd
choice = input("Do you want to print squares of even or odd numbers? (even/odd): ").lower() #This line ensures that any input from the user will be converted to lowercase.

# Check if the user chose even or odd, and calculate accordingly
if choice == 'even':
    print("\nSquares of even numbers between 100 and 200:")
    for number in range(100, 201):
        if number % 2 == 0:  # Check if the number is even
            square = number ** 2
            print(f"The square of {number} is {square}")
elif choice == 'odd':
    print("\nSquares of odd numbers between 100 and 200:")
    for number in range(101, 201):
        if number % 2 != 0:  # Check if the number is odd
            square = number ** 2
            print(f"The square of {number} is {square}")
else:
    print("Invalid choice! Please enter 'even' or 'odd'.")
```

#Exercise 4: Word counter

```python
def word_count(text):
    # Clean up punctuation from the text
    clean_text = ""
    for char in text:
        if char.isalnum() or char.isspace():  # Keep only alphanumeric characters and spaces
            clean_text += char
        else:
            clean_text += " "  # Replace punctuation with a space

    # Split the cleaned text into words
    words = clean_text.split()

    # Initialize an empty dictionary to store word counts
    word_freq = {}

    # Loop through each word in the text
    for word in words:
        # If the word is already in the dictionary, increment its count
        if word in word_freq:
            word_freq[word] += 1
        else:
            # Otherwise, add the word with a count of 1
            word_freq[word] = 1

    # Print each word and its count
    for word, count in word_freq.items():
        print(f"'{word}': {count}")

# Get input from the user
input_text = input("Enter your text: ").lower() #This line ensures that any input from the user will be converted to lowercase.
word_count(input_text)
```

#Exercise 5: Check for Palindrome

```python
str = input("Enter a string: ")# Take input from the user
clean_str = ''.join(char.lower() for char in str if char.isalnum())

if clean_str == clean_str[::-1]:# Check if the string is equal to its reverse
    print("True")
else:
    print("False")
```
